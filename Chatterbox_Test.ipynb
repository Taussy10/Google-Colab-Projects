```python

%%writefile /content/chatterbox_app.py

import torch
import torchaudio
import gradio as gr
import numpy as np
import random
import re
import traceback

from chatterbox.tts import ChatterboxTTS


# -----------------------------
# LOAD CHATTERBOX MODEL
# -----------------------------

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

print("Loading Chatterbox...")
model = ChatterboxTTS.from_pretrained(device=DEVICE)
print("Chatterbox loaded successfully on", DEVICE)


# -----------------------------
# RANDOM SEED
# -----------------------------

def set_seed(seed):
    if seed is not None and int(seed) != 0:
        seed = int(seed)

        torch.manual_seed(seed)

        if torch.cuda.is_available():
            torch.cuda.manual_seed_all(seed)

        np.random.seed(seed)
        random.seed(seed)


# -----------------------------
# SPLIT LONG TEXT
# -----------------------------

def split_text(text, max_chars=250):

    sentences = re.split(r'(?<=[.!?])\s+', text.strip())

    chunks = []
    current = ""

    for sentence in sentences:

        words = sentence.split()

        for word in words:

            candidate = (current + " " + word).strip()

            if len(candidate) <= max_chars:
                current = candidate

            else:
                if current:
                    chunks.append(current)

                current = word

        if current:
            chunks.append(current)
            current = ""

    return chunks


# -----------------------------
# GENERATE AUDIO
# -----------------------------

def generate_audio(
    text,
    reference_audio,
    exaggeration,
    cfg_weight,
    temperature,
    repetition_penalty,
    min_p,
    top_p,
    seed
):

    if not text or not text.strip():
        raise gr.Error("Please enter some text.")

    if reference_audio is None:
        raise gr.Error("Please upload a reference audio file.")

    try:

        set_seed(seed)

        chunks = split_text(text, max_chars=250)

        audio_parts = []

        print("\n========== GENERATION STARTED ==========")
        print("Total characters:", len(text))
        print("Total chunks:", len(chunks))

        for i, chunk in enumerate(chunks):

            print(f"\nGenerating chunk {i + 1}/{len(chunks)}")
            print("Text:", chunk)

            with torch.inference_mode():

                wav = model.generate(
                    chunk,
                    audio_prompt_path=reference_audio,
                    exaggeration=exaggeration,
                    cfg_weight=cfg_weight,
                    temperature=temperature,
                    repetition_penalty=repetition_penalty,
                    min_p=min_p,
                    top_p=top_p
                )

            audio = wav.squeeze().detach().cpu().float()

            audio_parts.append(audio)

            # Add a short pause between chunks
            if i < len(chunks) - 1:

                silence = torch.zeros(
                    int(model.sr * 0.25)
                )

                audio_parts.append(silence)

            if torch.cuda.is_available():
                torch.cuda.empty_cache()

        # Combine all generated audio
        final_audio = torch.cat(audio_parts)

        final_audio = final_audio.numpy()

        print("\nAll chunks generated successfully!")
        print("Final audio duration:",
              round(len(final_audio) / model.sr, 2), "seconds")

        print("========================================\n")

        return (model.sr, final_audio)

    except Exception as e:

        print("\n========== GENERATION ERROR ==========")
        traceback.print_exc()
        print("======================================")

        raise gr.Error(f"Generation failed: {str(e)}")


# -----------------------------
# GRADIO INTERFACE
# -----------------------------

with gr.Blocks(
    title="Chatterbox Voice Cloning Studio"
) as demo:

    gr.Markdown("# 🎙️ Chatterbox Voice Cloning Studio")

    gr.Markdown(
        "Upload a reference voice, enter your text, "
        "and customize the voice generation settings."
    )

    with gr.Row():

        with gr.Column(scale=1):

            text = gr.Textbox(
                label="Text to Speak",
                lines=5,
                placeholder="Enter the text you want to generate..."
            )

            reference_audio = gr.Audio(
                label="Reference Voice",
                sources=["upload", "microphone"],
                type="filepath"
            )

            with gr.Accordion(
                "Voice & Generation Settings",
                open=True
            ):

                exaggeration = gr.Slider(
                    minimum=0.25,
                    maximum=2.0,
                    value=0.5,
                    step=0.05,
                    label="Exaggeration"
                )

                cfg_weight = gr.Slider(
                    minimum=0.0,
                    maximum=1.0,
                    value=0.5,
                    step=0.05,
                    label="CFG Weight"
                )

                temperature = gr.Slider(
                    minimum=0.05,
                    maximum=5.0,
                    value=0.8,
                    step=0.05,
                    label="Temperature"
                )

                repetition_penalty = gr.Slider(
                    minimum=1.0,
                    maximum=2.0,
                    value=1.2,
                    step=0.05,
                    label="Repetition Penalty"
                )

                min_p = gr.Slider(
                    minimum=0.0,
                    maximum=1.0,
                    value=0.05,
                    step=0.01,
                    label="Min P"
                )

                top_p = gr.Slider(
                    minimum=0.0,
                    maximum=1.0,
                    value=1.0,
                    step=0.01,
                    label="Top P"
                )

                seed = gr.Number(
                    value=0,
                    precision=0,
                    label="Random Seed",
                    info="0 = random generation"
                )

            generate_btn = gr.Button(
                "Generate Speech",
                variant="primary"
            )

        with gr.Column(scale=1):

            output_audio = gr.Audio(
                label="Generated Voice",
                type="numpy"
            )

    generate_btn.click(
        fn=generate_audio,
        inputs=[
            text,
            reference_audio,
            exaggeration,
            cfg_weight,
            temperature,
            repetition_penalty,
            min_p,
            top_p,
            seed
        ],
        outputs=output_audio
    )


# -----------------------------
# LAUNCH
# -----------------------------

demo.queue(default_concurrency_limit=1).launch(
    share=True,
    server_name="0.0.0.0"
)
```
