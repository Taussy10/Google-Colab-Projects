```python
# Cell 2: Run the Voice Cloning Studio
import gradio as gr
from f5_tts.api import F5TTS

print("Loading Voice Cloning model on GPU...")
f5tts = F5TTS(device="cuda")

def clone(text, ref_audio, ref_text=""):
    if not ref_audio:
        return None
    wav, sr, _ = f5tts.infer(
        ref_file=ref_audio,
        ref_text=ref_text,
        gen_text=text
    )
    return sr, wav

demo = gr.Interface(
    fn=clone,
    inputs=[
        gr.Textbox(label="Text to Speak", lines=4, placeholder="Enter text to generate..."),
        gr.Audio(label="Reference Voice (Hemi / Any Audio)", type="filepath"),
        gr.Textbox(label="Reference Audio Transcript (Optional)", placeholder="Leave blank or type words from ref audio")
    ],
    outputs=gr.Audio(label="Cloned Voice Output"),
    title="Zero-Shot Voice Cloning Studio (GPU Accelerated)"
)

demo.launch(share=True)
```
