Do sentiment analysis

```python
# Import the pipeline function from Transformers -> to use the transformers
from transformers import pipeline

# Tell Transformers which task we want to perform.
# It automatically selects a suitable AI model.
#btw we can also provide model here by adding one more attribute -> "model"=""
task = pipeline("sentiment-analysis")

# Give input to the AI model
result = task("I love this product!")

# Display the output
print(result)
```
